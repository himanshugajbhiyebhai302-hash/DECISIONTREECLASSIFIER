```python
from sklearn import datasets
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
```

```python
from sklearn import datasets
import pandas as pd # Ensure pandas is imported if not already

iris = datasets.load_iris()
iris_frame  = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_frame.head()
```

```python
iris.train,iris.test,iris.target_train,iris.target_test = train_test_split(iris_frame, iris.target,train_size = 0.2, random_state=42)
```

```python
dt = DecisionTreeClassifier(max_depth = 6 , criterion = 'gini')
model = dt.fit(iris.train,iris.target_train)
```

```python
predictedData = model.predict(iris.test)
```

```python
from sklearn.tree import export_graphviz
dot_data_tree = export_graphviz(model, out_file=None,
                                    feature_names=iris.feature_names,
                                    class_names=iris.target_names,
                                    filled=True, rounded=True,
                                    special_characters=True)
```

```python
import graphviz
graph = graphviz.Source(dot_data_tree)
graph
```

```python
from sklearn.tree import export_text
r = export_text(model, feature_names = iris.feature_names)
print(r)
```
