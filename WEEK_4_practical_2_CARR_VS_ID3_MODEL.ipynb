```python
# 1.IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.metrics import accuracy_score, classification_report
```

```python
# 2. CREATE A SAMPLE DATASET
print("="*50)
data = {
    "study_Hours" : [1,2,3,4,5,6,7],
    "Attendence"  : [20,10,30,40,89,95,78],
    "Pass_Fail"   : [0,0,0,1,1,1,1]
}
print(data)
```

```python
import pandas as pd

# =================================
# 3. Separate input(x) and output(y)
# =================================

# Convert the dictionary to a DataFrame to easily select multiple columns
df = pd.DataFrame(data)
x = df[['study_Hours', 'Attendence']] # Correctly select multiple columns for x

# Setting 'Pass_Fail' as the target variable for y
y = df[['Pass_Fail']]

print('\n , x : ',x)
print("\n, y :",y)
```

```python
# 4.Train test split
x_train, x_test, y_train, y_test = train_test_split(
    x ,
    y,
    test_size = 0.3 ,
    random_state = 42)
print("\nTraining Features (x_train):")
print(x_train)
print("\nTraining Target (y_train):")
print(y_train)
print("\nTesting Features (x_test):")
print(x_test)
print("\nTesting Target (y_test):")
print(y_test)
```

```python
# ===============================================
# 5. CONFIRMING THE MODEL (now using a Classifier)
# ===============================================

# Note : CART BASED DECISION TREE KO => GINI, GINI can predict binary encoding
# NOTE : ID3 BASED DECISION TREE KO => ENTROPHY, it can predict multiple classes

dt_classifier = DecisionTreeClassifier(max_depth=3, criterion ='gini', random_state=42) # Changed to Classifier
model = dt_classifier.fit(x_train, y_train)
```

```python
# =============
# 6. PREDICTION
# =============

y_pred = model.predict(x_test)
print("\n Actual data (first 5 samples of training features):\n", x_train.head())
print("\nPrediction for x_test:\n", y_pred)

# ======================================
# 7. Evaluate with classification metrics
# =======================================

accuracy = accuracy_score(y_test, y_pred)
print(f"\nAccuracy: {accuracy:.2f}")

# ================================
# 8. Evaluate classification Report
# =================================

print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
from sklearn.tree import plot_tree

# ==========
# 9 . FIGURE
# ==========

# samples = This shows the rules learned by the model.
# gini    = Impurity of the node
# ENTROPY = PREDICTED class

plt.figure(figsize=(15, 10))
plot_tree(
    model,
    feature_names=x.columns.tolist(),
    class_names=['Fail', 'Pass'],
    filled=True,
    fontsize = 12,
    rounded=True
    )
plt.title('Decision Tree Classifier Structure')
plt.show()
```

```python
# =======================
# 10 PREDICT A NEW STUDENT
#========================
# New student :
# study hours = 4
# Attendence = 75%

# Here we are using new case where we add new four entries with reverse
# pass fail ratio would be different on first half we have more no pass
# here if the dimension of prediction == 1 ; pass
new_student = [[4,75]]
prediction = model.predict(new_student)
if prediction[0] == 1:
  print(f"\n New Student Prediction : Pass")
else:
  print(f"\n New Student Prediction : Fail")
```

```python
# ==========================
# 12 Predict Probability
# ===========================
probability = model.predict_proba(new_student)
print("\nProbability:")
print("\n Fail:",probability[0][0])
print("\n Pass:",probability[0][1])

```

```python
# ==========================
# 13. FIND Feature importance
# ==========================

# Feature importance tells us which variable.
# contributed more to the decision tree.

importance = pd.DataFrame({
    "Feature ": x.columns,
    "Importance": model.feature_importances_
})

print("\nFeature Importance")
print(importance)

```

```python
# =================
# 14. Compare Cart and ID3-STYLE TREE
# =================

# cart --> Gini Impurity
cart_model = DecisionTreeClassifier(
    criterion = 'gini',
    max_depth = 3,
    random_state = 42
)
cart_model.fit(x_train, y_train)

# ID3 based model --> Entrophy
ID3_model = DecisionTreeClassifier(
    criterion = 'entropy',
    max_depth = 3,
    random_state = 42
)
ID3_model.fit(x_train, y_train)

# PREDICTION BETWEEN BOTH MODELS
cart_prediction = cart_model.predict(x_test)
ID3_model_prediction = ID3_model.predict(x_test)

print("\nCart Model Prediction:")
print("accuracy_score:",accuracy_score(y_test, cart_prediction))

print("\nID3 Model Prediction:")
print("accuracy_score:",accuracy_score(y_test, ID3_model_prediction))
```
